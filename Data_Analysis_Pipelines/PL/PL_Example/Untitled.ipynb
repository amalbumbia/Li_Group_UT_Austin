```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
import plotly.graph_objects as go
import pickle
%matplotlib qt
```

```python
#Parameters:
suffix = '_1.txt'
bkg_file = 'bare.txt'
# imgs arranged as [bkg, 1 , 2, 3 ...]
imgs = Path('').glob('*'+suffix)
# for path in imgs:
#     print(path)
low_wl = 600 # nm
high_wl = 675 # nm
```

```python
# Helper function to load spec files, NOT STANDALONE relies on parameters defined above
def load_spec(file):
    if Path(file).suffix == '.txt':
        array = np.loadtxt(file, delimiter=',')
        array = np.delete(array, 1, 1)
    return array
```

```python
bkg = load_spec(bkg_file)
```

```python
spec_array=[]
labels = []

for path in imgs:
    print(path)
    array = load_spec(path)
#     array[:,1] = (array[:,1]-bkg[:,1])/bkg[:,1]
    spec_array.append(array)
    labels.append(str(path).replace(suffix,''))
```

```python
fig=go.Figure()

for i, spec in enumerate(spec_array):
    low_ind = np.argmin(abs(spec[:,0]-low_wl))
    high_ind = np.argmin(abs(spec[:,0]-high_wl))
    fig.add_trace(go.Scatter(x= spec[low_ind:high_ind,0], y =spec[low_ind:high_ind,1],
                            name=labels[i]
                            ))
    fig.write_html("bright_spot_spectra.html")
fig.show()
```
